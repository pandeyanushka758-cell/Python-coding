# Project 1: Electricity Bill Calculator

Create a Python program that calculates an electricity bill based on the number of units consumed.

### Billing Rules

| Units Consumed  |     Rate |
| --------------- | -------: |
| 0–100 units     |  ₹3/unit |
| 101–200 units   |  ₹5/unit |
| 201–500 units   |  ₹7/unit |
| Above 500 units | ₹10/unit |

The billing should be **slab-based**, meaning each portion of the consumption is charged at its corresponding rate.

For example, if consumption is **250 units**:

```text
First 100 × ₹3  = ₹300
Next 100 × ₹5   = ₹500
Next 50 × ₹7    = ₹350
                  ----
Total            = ₹1150
```

### Additional Charges

After calculating the energy charge:

* If the bill is above ₹2,000 → add **5% surcharge**
* If the bill is above ₹5,000 → add **10% surcharge**
* Otherwise → no surcharge

### Input

Take:

```text
Customer Name
Units Consumed
```

### Sample Input

```text
Customer Name: Rahul
Units Consumed: 250
```

### Sample Output

```text
========== ELECTRICITY BILL ==========

Customer: Rahul
Units Consumed: 250

Energy Charge: ₹1150
Surcharge: ₹0
--------------------------------
Total Bill: ₹1150
```

### Another Sample

**Input**

```text
Customer Name: Amit
Units Consumed: 650
```

**Output**

```text
========== ELECTRICITY BILL ==========

Customer: Amit
Units Consumed: 650

Energy Charge: ₹4650
Surcharge: ₹232.50
--------------------------------
Total Bill: ₹4882.50
```

### Validation

Your program should handle:

```text
Units = -50
```

Output:

```text
Invalid units
```

### Restrictions

Try solving it using:

* `if`
* `elif`
* `else`
* arithmetic operators

**Don't use functions, lists, dictionaries, or loops initially.**

# Project 2: Income Tax Calculator

Create a Python program that calculates income tax based on annual income.

Use the following **simplified tax slabs for practice**:

| Annual Income           | Tax Rate |
| ----------------------- | -------: |
| Up to ₹4,00,000         |       0% |
| ₹4,00,001 – ₹8,00,000   |       5% |
| ₹8,00,001 – ₹12,00,000  |      10% |
| ₹12,00,001 – ₹16,00,000 |      15% |
| ₹16,00,001 – ₹20,00,000 |      20% |
| Above ₹20,00,000        |      30% |

Use **slab-based taxation**.

For example, for an income of ₹10,00,000:

```text
First ₹4,00,000       → 0%
Next ₹4,00,000        → 5%
Remaining ₹2,00,000   → 10%
```

Therefore:

```text
Tax = ₹0 + ₹20,000 + ₹20,000
Tax = ₹40,000
```

### Input

Take:

```text
Name
Annual Income
```

### Sample Input

```text
Name: Rahul
Annual Income: 1000000
```

### Sample Output

```text
========== INCOME TAX CALCULATOR ==========

Name: Rahul
Annual Income: ₹1000000

Taxable Income: ₹1000000
Income Tax: ₹40000
--------------------------------
Final Income After Tax: ₹960000
```

### Another Sample

**Input**

```text
Name: Priya
Annual Income: 1800000
```

**Output**

```text
========== INCOME TAX CALCULATOR ==========

Name: Priya
Annual Income: ₹1800000

Income Tax: ₹160000
--------------------------------
Final Income After Tax: ₹1640000
```

### Validation

If the user enters:

```text
Annual Income: -50000
```

Output:

```text
Invalid income
```
