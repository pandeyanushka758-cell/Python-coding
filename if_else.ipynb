```python
def greet (name):
    import datetime
    h=datetime.datetime.now().hour
    if h>=18:
        print(f'good evening {name.upper()}')
    elif h<18 and h>12:
        print(f'good afternoon {name.upper()}')
    else:
        print(f"good morning {name.upper()}")
greet("bulbul")
```
