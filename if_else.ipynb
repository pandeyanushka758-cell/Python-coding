```python
#20 Simple Calculator

num1 = int(input("Enter first number "))
operator = input("Enter operator (+, -, *, /) ")
num2 = int(input("Enter second number: "))

if operator == "+":
    print("Ans:", num1 + num2)

elif operator == "-":
    print("Ans:", num1 - num2)

elif operator == "*":
    print("Ans:", num1 * num2)

elif operator == "/":
    if num2 != 0:
        print("Ans:", num1 / num2)
    else:
        print("Cannot divide by zero")

else:
    print("Invalid operator")
```
